## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

url = "https://huggingface.co/datasets/FlyRank/internship-starter/resolve/main/content_refresh_anonymized.csv"
df = pd.read_csv(url)

print(f"Dataset size: {df.shape[0]:,} rows × {df.shape[1]} columns")

print("\nContent age (days):")
print(df["content_age_days"].describe()[["mean", "50%", "min", "max"]])

print("\nTrend distribution:")
trend_counts = df["trend_direction"].value_counts()
trend_pct = (df["trend_direction"].value_counts(normalize=True) * 100).round(1)

print(pd.DataFrame({
    "count": trend_counts,
    "percentage": trend_pct
}))

print("\nMissing values in key fields:")
key_fields = ["content_id", "content_age_days", "trend_direction"]
print(df[key_fields].isna().sum())

display(df.head(3))
```
